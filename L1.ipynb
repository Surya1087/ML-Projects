```python
# Import TensorFlow for deep learning
import tensorflow as tf

# Import Keras for neural network development
from tensorflow import keras

# Import PyTorch for deep learning
import torch

# Import NumPy for numerical operations
import numpy as np

# Import Pandas for data handling
import pandas as pd

# Import Matplotlib for visualization
import matplotlib.pyplot as plt

# Import package version utility
import importlib.metadata as metadata

# Display successful import message
print("All libraries imported successfully!")
```

# TensorFlow Implementation

```python
# Create two constant tensors
a = tf.constant([10, 20, 30])
b = tf.constant([1, 2, 3])

# Perform addition of tensors
c = tf.add(a, b)

# Display the tensors
print("Tensor A:", a.numpy())
print("Tensor B:", b.numpy())

# Display addition result
print("Addition:", c.numpy())

# Perform multiplication
print("Multiplication:", tf.multiply(a, b).numpy())
```

# Keras implementation

```python
# Import required Keras layers
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Create a sequential neural network
model = Sequential()

# Add first hidden layer with 64 neurons
model.add(Dense(64, activation='relu', input_shape=(10,)))

# Add second hidden layer with 32 neurons
model.add(Dense(32, activation='relu'))

# Add output layer with one neuron
model.add(Dense(1, activation='sigmoid'))

# Display model architecture
model.summary()
```

# PyTorch implementation

```python
# Create two PyTorch tensors
a = torch.tensor([10, 20, 30])
b = torch.tensor([1, 2, 3])

# Perform tensor addition
c = torch.add(a, b)

# Display tensors
print("Tensor A:", a)
print("Tensor B:", b)

# Display addition result
print("Addition:", c)

# Perform multiplication
print("Multiplication:", torch.mul(a, b))
```
