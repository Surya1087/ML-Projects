# Import required libraries

```python
# Import TensorFlow
import tensorflow as tf

# Import Keras
from tensorflow import keras

# Import required layers
from tensorflow.keras import layers

# Import NumPy
import numpy as np

# Import Matplotlib
import matplotlib.pyplot as plt

# Import classification metrics
from sklearn.metrics import classification_report, confusion_matrix
```

# Load CIFAR-10 dataset

```python
# Load CIFAR-10 training and testing data
(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()

# Display training image shape
print("Training Images:", x_train.shape)

# Display training labels shape
print("Training Labels:", y_train.shape)

# Display testing image shape
print("Testing Images:", x_test.shape)

# Display testing labels shape
print("Testing Labels:", y_test.shape)

# Display total training images
print("Total Training Images:", len(x_train))

# Display total testing images
print("Total Testing Images:", len(x_test))
```

# Define CIFAR-10 class names

```python
# Define the ten image categories
class_names = [
    'Airplane',
    'Automobile',
    'Bird',
    'Cat',
    'Deer',
    'Dog',
    'Frog',
    'Horse',
    'Ship',
    'Truck'
]

# Display class names
print("CIFAR-10 Classes:")

for i, name in enumerate(class_names):
    print(i, ":", name)
```

# Display sample images

```python
# Create figure
plt.figure(figsize=(12, 6))

# Display first 10 training images
for i in range(10):

    # Create subplot
    plt.subplot(2, 5, i + 1)

    # Display image
    plt.imshow(x_train[i])

    # Display image class
    plt.title(class_names[int(y_train[i][0])])

    # Hide axis
    plt.axis('off')

# Adjust spacing
plt.tight_layout()

# Display images
plt.show()
```

# Preprocess image data

```python
# Normalize training images between 0 and 1
x_train = x_train.astype('float32') / 255.0

# Normalize testing images
x_test = x_test.astype('float32') / 255.0

# Convert labels into one-dimensional arrays
y_train = y_train.flatten()
y_test = y_test.flatten()

# Display processed image shape
print("Training Data Shape:", x_train.shape)

# Display pixel value range
print("Minimum Pixel Value:", x_train.min())
print("Maximum Pixel Value:", x_train.max())

# Display successful message
print("Image preprocessing completed!")
```

# Define CNN model architecture

```python
# Create sequential CNN model
model = keras.Sequential([

    # Define input image shape
    keras.Input(shape=(32, 32, 3)),

    # First convolution layer with 32 filters
    layers.Conv2D(32, (3, 3), activation='relu', padding='same'),

    # Reduce image dimensions
    layers.MaxPooling2D((2, 2)),

    # Second convolution layer with 64 filters
    layers.Conv2D(64, (3, 3), activation='relu', padding='same'),

    # Reduce image dimensions
    layers.MaxPooling2D((2, 2)),

    # Third convolution layer with 128 filters
    layers.Conv2D(128, (3, 3), activation='relu', padding='same'),

    # Flatten feature maps into one-dimensional vector
    layers.Flatten(),

    # Fully connected layer with 128 neurons
    layers.Dense(128, activation='relu'),

    # Dropout to reduce overfitting
    layers.Dropout(0.5),

    # Output layer with 10 classes
    layers.Dense(10, activation='softmax')

])

# Display model architecture
model.summary()
```

# Compile the CNN model

```python
# Compile the CNN model
model.compile(

    # Use Adam optimizer
    optimizer='adam',

    # Use sparse categorical crossentropy
    loss='sparse_categorical_crossentropy',

    # Measure classification accuracy
    metrics=['accuracy']

)

# Display compilation message
print("CNN Model compiled successfully!")
```

# Train the CNN model

```python
# Train the CNN model
history = model.fit(

    # Training images
    x_train,

    # Training labels
    y_train,

    # Number of epochs
    epochs=10,

    # Number of images processed per batch
    batch_size=64,

    # Use 20% training data for validation
    validation_split=0.2,

    # Display training progress
    verbose=1

)

# Display completion message
print("CNN Model training completed!")
```

# Evaluate model performance

```python
# Evaluate model using testing data
test_loss, test_accuracy = model.evaluate(x_test, y_test)

# Display testing loss
print("Test Loss:", test_loss)

# Display testing accuracy
print("Test Accuracy:", test_accuracy * 100, "%")
```

# Plot training and validation accuracy

```python
# Create accuracy graph
plt.figure(figsize=(8, 5))

# Plot training accuracy
plt.plot(history.history['accuracy'], label='Training Accuracy')

# Plot validation accuracy
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')

# Add graph title
plt.title('CNN Training and Validation Accuracy')

# Add X-axis label
plt.xlabel('Epochs')

# Add Y-axis label
plt.ylabel('Accuracy')

# Display legend
plt.legend()

# Display grid
plt.grid(True)

# Show graph
plt.show()
```

# Plot training and validation loss

```python
# Create loss graph
plt.figure(figsize=(8, 5))

# Plot training loss
plt.plot(history.history['loss'], label='Training Loss')

# Plot validation loss
plt.plot(history.history['val_loss'], label='Validation Loss')

# Add graph title
plt.title('CNN Training and Validation Loss')

# Add X-axis label
plt.xlabel('Epochs')

# Add Y-axis label
plt.ylabel('Loss')

# Display legend
plt.legend()

# Display grid
plt.grid(True)

# Show graph
plt.show()
```

# Predict image classes

```python
# Predict first 10 testing images
predictions = model.predict(x_test[:10])

# Find class with highest probability
predicted_labels = np.argmax(predictions, axis=1)

# Display actual and predicted labels
for i in range(10):

    print(
        "Actual:", class_names[y_test[i]],
        "| Predicted:", class_names[predicted_labels[i]]
    )
```

# Display predicted images

```python
# Create figure
plt.figure(figsize=(12, 6))

# Display first 10 testing images
for i in range(10):

    # Create subplot
    plt.subplot(2, 5, i + 1)

    # Display image
    plt.imshow(x_test[i])

    # Get actual and predicted class names
    actual = class_names[y_test[i]]
    predicted = class_names[predicted_labels[i]]

    # Display prediction
    plt.title(
        f"Actual: {actual}\nPredicted: {predicted}"
    )

    # Hide axis
    plt.axis('off')

# Adjust layout
plt.tight_layout()

# Display output
plt.show()
```

# Classification report

```python
# Predict all testing images
all_predictions = model.predict(x_test)

# Get predicted class labels
y_pred = np.argmax(all_predictions, axis=1)

# Generate classification report
print(classification_report(
    y_test,
    y_pred,
    target_names=class_names
))
```

# Display confusion matrix

```python
# Calculate confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Display confusion matrix
plt.figure(figsize=(10, 8))

# Create heatmap
plt.imshow(cm, cmap='Blues')

# Add color bar
plt.colorbar()

# Display class names on axes
plt.xticks(range(10), class_names, rotation=45)
plt.yticks(range(10), class_names)

# Add axis labels
plt.xlabel('Predicted Class')
plt.ylabel('Actual Class')

# Add title
plt.title('CIFAR-10 Confusion Matrix')

# Display graph
plt.tight_layout()
plt.show()
```
